## Limits to plan around

As discussed and not yet verified: a Colab session with a T4 GPU runs at most about 12 hours and stops after about 90 minutes idle, and GPU access is rationed by recent use. Work that only calls an API needs no GPU, so run it on the CPU runtime; keep files on a mounted Drive so a disconnect loses nothing. Piping a script from the internet into a shell runs whatever that address serves: read it first.

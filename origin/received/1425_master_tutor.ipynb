### 📸 Stuck? Let Ghonchu AI Check Your Mechanical Layout
Run the cell below on your phone. It will open your phone's **rear-facing camera** so you can take a clear picture of your physical chassis and show it to the AI.

```python
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

def activate_ghonchu_eye(filename='chassis_check.jpg', quality=0.8):
  js = Javascript('''
    async function captureImage(quality) {
      const div = document.createElement('div');
      const button = document.createElement('button');
      button.textContent = '📸 Snap Photo for Ghonchu AI';
      button.style = 'padding:15px; background:#0284c7; color:white; border:none; border-radius:8px; font-weight:bold; font-size:16px; margin-top:10px; cursor:pointer;';
      
      const video = document.createElement('video');
      video.style.display = 'block';
      video.style.borderRadius = '8px';
      video.style.width = '100%';
      video.style.maxWidth = '400px';
      
      // Targets the physical environment (rear) camera on mobile browsers
      const stream = await navigator.mediaDevices.getUserMedia({
        video: { facingMode: 'environment' }
      });

      document.body.appendChild(div);
      div.appendChild(video);
      div.appendChild(button);
      video.srcObject = stream;
      await video.play();

      return new Promise((resolve) => {
        button.onclick = () => {
          const canvas = document.createElement('canvas');
          canvas.width = video.videoWidth;
          canvas.height = video.videoHeight;
          canvas.getContext('2d').drawImage(video, 0, 0);
          stream.getVideoTracks().stop();
          div.remove();
          resolve(canvas.toDataURL('image/jpeg', quality));
        };
      });
    }
  ''')
  display(js)
  data = eval_js('captureImage({})'.format(quality))
  binary = b64decode(data.split(',')[1])
  with open(filename, 'wb') as f:
    f.write(binary)
  return filename

try:
  filename = activate_ghonchu_eye()
  print('Captured! Saved as {} - Upload this to the Ghonchu Chat Panel on the right for inspection.'.format(filename))
except Exception as e:
  print('Camera initialization error:', str(e))
```

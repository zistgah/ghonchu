### 📸 Step 2: Activate Ghonchu Eye (Mobile Rear Camera Integration)
If students encounter structural execution bugs, missing components, or alignment errors on the chassis base, running this cell enables the browser layer to interface directly with the mobile device's rear hardware array.

```python
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

def capture_hardware_alignment(filename='chassis_layout.jpg', quality=0.8):
  js = Javascript('''
    async function captureMobileFrame(quality) {
      const wrapper = document.createElement('div');
      const btn = document.createElement('button');
      btn.textContent = '📸 Analyze Layout with Ghonchu AI';
      btn.style = 'padding:14px; background:#0284c7; color:white; border:none; border-radius:6px; font-weight:bold; cursor:pointer;';
      
      const vid = document.createElement('video');
      vid.style.display = 'block';
      vid.style.borderRadius = '8px';
      vid.style.width = '100%';
      vid.style.maxWidth = '480px';
      
      // Constrains browser execution context to the phone back camera lens
      const stream = await navigator.mediaDevices.getUserMedia({
        video: { facingMode: 'environment' }
      });

      document.body.appendChild(wrapper);
      wrapper.appendChild(vid);
      wrapper.appendChild(btn);
      vid.srcObject = stream;
      await vid.play();

      return new Promise((resolve) => {
        btn.onclick = () => {
          const canvas = document.createElement('canvas');
          canvas.width = vid.videoWidth;
          canvas.height = vid.videoHeight;
          canvas.getContext('2d').drawImage(vid, 0, 0);
          stream.getVideoTracks().stop();
          wrapper.remove();
          resolve(canvas.toDataURL('image/jpeg', quality));
        };
      });
    }
  ''')
  display(js)
  data = eval_js('captureMobileFrame({})'.format(quality))
  raw_bytes = b64decode(data.split(','))[1]
  with open(filename, 'wb') as f:
    f.write(raw_bytes)
  return filename

try:
  target_frame = capture_hardware_alignment()
  print('Frame verified. Target array saved directly to disk buffer: ' + target_frame)
except Exception as e:
  print('Hardware allocation exception: ', str(e))
```
